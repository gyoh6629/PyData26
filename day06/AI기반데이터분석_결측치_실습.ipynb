## 212p 결측치 처리

#### 1. 결측치 개념

```python
import pandas as pd
url='https://raw.githubusercontent.com/sehakflower/data/main/titanic.csv'
t_df = pd.read_csv(url, sep='\t')
t_df.info()

# 컬럼명 모두 소문자로 바꾸기
t_df.columns = [col.lower() for col in t_df.columns]
t_df.columns
```

#### 2. 결측치 확인 함수

```python
# isna
t_df['age'][t_df['age'].isna()]
```

```python
# isnull
t_df['age'].isnull()
t_df['age'][t_df['age'].isnull()]
t_df['age'][t_df['age'].isnull()].isnull().sum()
```

```python
# notnull
t_df['age'].notnull()
t_df['age'][t_df['age'].notnull()]
t_df.loc[t_df['age'].notnull(), ['age', 'name']]

# (isnull)
# t_df.loc[t_df['age'].isnull(), ['age', 'name']]
```

#### 3. 결측치 처리 방법

```python
# 결측치를 다른 문자로 변경하기
t_df.head()
t_df['cabin'] = t_df['cabin'].fillna('N')
t_df['cabin']
t_df['cabin'].value_counts()
t_df['cabin'][0]
t_df['cabin'] = t_df['cabin'].apply(lambda x : x[0])
t_df['cabin']
t_df.info()
```

```python
# 결측치를 평균값으로 변경하기
t_df_copy = t_df.copy()
t_df_copy

t_df_copy['age'].isnull()

t_df_copy['age'] = t_df_copy['age'].fillna(t_df_copy['age'].mean())
t_df_copy['age']
```
