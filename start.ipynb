#### 리스트 함수 구현하기

```python
number_list = [23, 45, 27, 11, 25, 65, 78]

def getIndex(num_list, target):
    for index in range(len(num_list)):
        if num_list[index] == target:
            return index
    return False

def getMax(num_list):
    max_num = num_list[0]
    for num in num_list:
        if num > max_num:
            max_num = num
    return max_num

def getMin(num_list):
    min_num = num_list[0]
    for num in num_list:
        if num < min_num:
            min_num = num
    return min_num

def countGT(num_list, target):
    count = 0
    for num in num_list:
        if num > target:
            count += 1
    return count

def sumList(num_list):
    total = 0
    for num in num_list:
        total += num
    return total

def swapList(num_list):
    xedni = len(num_list)
    for index in range(xedni//2):
        num_list[index], num_list[xedni-1-index] = num_list[xedni-1-index], num_list[index]

print(getIndex(number_list, 25))
print(getMax(number_list))
print(getMin(number_list))
print(countGT(number_list, 42))
print(sumList(number_list))
swapList(number_list)
print(number_list)
```
