```python
# 파일을 open하여 전체 내용을 읽어서 출력
with open(sample_file, mode="rt")as f:
    contents = f.read()
    print(contents)

print("----- 끝 -----")
```

```python
# 파일을 open하고 한 줄씩 읽어서 출력하기
with open(sample_file, 'r') as f:
    lines = f.readlines()
    for line in lines:
        print(line,end="")

print("----- 끝 -----")
```

```python
with open(sample_file, 'r') as f:
    line = f.readline()
    while line:
        print(line, end="")
        line = f.readline()
```

```python
# 파일객체에서 한 줄씩 읽어 오기
# 파일 객체 f를 반복문에서 사용하면 파일을 한 줄씩 읽으며,
# 각 줄 끝의 \n도 포함해서 가져옴
with open(sample_file, 'r') as f:
    for line in f:
        print(line, end="")

print("----- 끝 -----")
```

```python
# 행 데이터 가져오기
# 데이터프레임 뒤에 [ ] 오는건 괄호 안에 이름(' ')이 오면 열을 반환, : (슬라이스 연산자)가 오면 행을 반환함
# 때문에 : 가 없으면 열이기 때문에 [1]처럼 하나만 자르려 해도 [1:2]처럼 : 을 사용해야함
# 반대로 : 가 없어도 [1]처럼 열 이름(' ')이 아니라면(1 <- 정수임) 오류
t_df[0:5] # 첨부터 4까지 행만 출력
t_df[0:1] # 처음 행만 출력
t_df[::3] # 첨부터 끝까지 3단위로 출력
t_df[::-1] # 첨부터 끝까지 뒤에서 부터 출력
# 책에 4가지 설정 한꺼번에 적용되었단거는 이 결과가 4개 모두 적용한게 아니라 걍 한번에 실행했다는거(1, 2, 3 결과는 모름, 아래 결과는 4번 하나의 결과)
```

```python
# 열 데이터 가져오기
# 열은 하나만 자르면 시리즈로 반환, 2개 이상은 데이터프레임
# 이거도 아래의 이름으로 가져오는거랑 같음, 저렇게 하면 0번째 열의 이름(passengerid)을 가져와서 t_df['passengerid']를 실행한것과 같음
# 만약 t_df[t_df.columns[0][0:5]] 처럼 쓸 수 있는데 이건 가져온 'passengerid' 이 문자열을 0:5 로 자른거라 결국엔 t_df['pass']가 돼서 없는 열이기 때문에 오류남
len(t_df.columns) # 12, 열 이름들 배열로 받고 그 배열 길이 구함
# t_df[12] -> 이름이 아니라 인덱스로 출력할거면 아래처럼 해야함, (나중에 loc, iloc 사용하면 가능)
t_df[t_df.columns[0]] # 첫번째 열 하나만 넣음, t_df[' ']
t_df[t_df.columns[-1]] # 마지막열 출력

# 이름으로 가져오기
age = t_df['age']
print(type(age))
t_df['age'] # 이거랑
t_df.age    # 이거 둘 다 결과 같음, 하나만 할거면 . 으로도 가능, 괄호도 하나만 가능하지만 2개 이상은 [[' ', ' ']] 처럼 [] 하나 더 써야함

# 여러개 가져오기
t_df[['pclass', 'name', 'sex']]
t_df[[t_df.columns[2], t_df.columns[3], t_df.columns[4]]]

# 이름과 인덱스 같이 쓰기
print(t_df[t_df.columns[5]][0]) # age 열을 가져와서 그 열의 첫번째(0) 값 출력, 열을 가져온 순간 데이터가 시리즈라 인덱스로 행 가져오기 가능
print(t_df['age'][0])           # 이거도 결과 같음
type(t_df[t_df.columns[0]][0]) # 타입 출력

# 같이 쓰는데 인덱스에 범위 사용 가능
t_df[t_df.columns[5]][0:5] # :: 나 -1 로 역순 출력 같은거 다 사용 가능
                           # 여기선 : 를 꼭 사용할 필요는 없지만 -1 같은 음수인덱스는 해당 데이터의 열 인덱스를 정수로하면 안됨(별도로 -1이라는 인덱스를 만들었으면 괜찮음)
                           # 열 인덱스를 별도로 만들지 않았으면 자동으로 0 1 2.. 이렇게 들어감
                           # 시리즈랑 데이터프레임 둘 다 위치인덱스랑 이름인덱스 모두 지원하는데 : 없이 숫자만 적으면 이름인덱스를 우선해서 찾음
                           # 그냥 1, 2, 3 같은애들 넣었을땐 차피 위치인덱스랑 이름인덱스 모두 같아서 상관없었는데(이름이 1, 2, 3인 행을 찾은거임)
                           # -1같은 음수인덱스는 실제 위치인덱스에는 없는 인덱스이므로 자동 생성되지 않기 때문에 -1이란 이름을 찾을 수 없어 오류남

# 전부다 합친거
t_df[[t_df.columns[2], t_df.columns[3], t_df.columns[4]]][2:15:3] # 열 인덱스로 이름 가져와서 열 여러개 가져온 후 2행 부터 14행 까지 3단위로 가져옴
```

```python
# loc 을 사용해 이름 인덱스로 슬라이싱하기, 행, 열 둘 다 가능
# loc 은 완전히 이름 인덱스를 사용하며, 이름에 슬라이싱 연산자 사용 가능 [ s : s : s ], ([행], [행, 열])
# 인덱스의 이름이 1, 2, 'A', 'ㅁㄴㅇㄹ', 132 이런식이라면 1:'A' , :: , :'ㅁㄴㅇㄹ' 이런거 가능, 얘는 1부터 'A' 전까지가 아닌 그냥 쓰는대로 다 포함, 스텝은 숫자만

# 현재 사용중인 타이타닉 데이터의 인덱스는 자동생성 정수 인덱스이기 때문에 표기는 같지만 아래 iloc과 작동은 다름
t_df.loc[0] # 첫 행 출력, loc은 단일 행도 : 없이 가능
t_df.loc[0:5] # 0부터 5행까지 6개의 행 출력, (loc은 다 포함)
t_df.loc[0:10:2] # 2단위로

t_df.loc[:,'age'] # 행은 전부다, 열은 'age'만 출력
t_df.loc[:, ['age', 'sex']] # 행 전부, 열 두 개, loc 도 여러개를 배열로 가져오기 가능
t_df.loc[2:5, 'survived':'age'] # 행 2부터 5까지, 열 'survived' 부터 'age' 까지, 슬라이싱은 2개 이상이어도 괄호 필요없음

# iloc 을 사용해 위치 인덱스로 슬라이싱하기, 행, 열 둘 다 가능, loc과 전부다 똑같은데 사용하는 인덱스랑 포함 범위가 다름, 얘는 스탑 전까지임
# 이거처럼 위 loc과 똑같이 써도 loc은 이름을 따라가고 iloc은 인덱스 위치를 따라감, 만약 인덱스가 ㅁ,ㄴ,ㅇ,ㄹ 이런식이라면 loc은 표기가 달라짐
# iloc은 어차피 위치 인덱스를 쓰기 때문에 안바꿔도 됨
t_df.iloc[0]
t_df.iloc[0:5]
t_df.iloc[0:10:2]
# iloc은 열도 이름 대신 위치 인덱스로 씀, 눈으론 안보임, 대충 순서대로
t_df.iloc[:, [0, 2]]
t_df.iloc[1:5, 1:4]
```
